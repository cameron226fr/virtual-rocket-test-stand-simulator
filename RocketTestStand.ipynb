```python
!pip install -q numpy pandas matplotlib ipywidgets

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time
import random

from datetime import datetime
from IPython.display import display, clear_output
import ipywidgets as widgets

print("Rocket Test Stand Simulator")
print("Python environment initialized.")
print("Dependencies loaded successfully.")

class RocketTestStand:

    def __init__(self):
        self.status = "OFFLINE"
        self.phase = "IDLE"
        self.elapsed_time = 0

        self.sensors = {
            "Chamber Pressure": 0.0,
            "Engine Temperature": 25.0,
            "Fuel Flow": 0.0,
            "Oxidizer Flow": 0.0,
            "Vibration": 0.0,
            "Tank Pressure": 0.0
        }

        self.telemetry = []
        self.events = []

    def log_event(self, message):
        timestamp = datetime.now().strftime("%H:%M:%S")

        self.events.append({
            "Timestamp": timestamp,
            "Event": message
        })

    def update_sensors(self, readings):
        self.sensors.update(readings)

    def record_telemetry(self):
        reading = {
            "Time": self.elapsed_time,
            "Timestamp": datetime.now().strftime("%H:%M:%S"),
            **self.sensors
        }

        self.telemetry.append(reading)

    def get_telemetry(self):
        return pd.DataFrame(self.telemetry)

    def get_event_log(self):
        return pd.DataFrame(self.events)

    def reset(self):
        self.status = "READY"
        self.phase = "IDLE"
        self.elapsed_time = 0

        self.sensors = {
            "Chamber Pressure": 0.0,
            "Engine Temperature": 25.0,
            "Fuel Flow": 0.0,
            "Oxidizer Flow": 0.0,
            "Vibration": 0.0,
            "Tank Pressure": 0.0
        }

        self.telemetry.clear()
        self.events.clear()

        self.log_event("Simulation reset.")

        print("Simulation reset successfully.")


test_stand = RocketTestStand()

test_stand.status = "READY"
test_stand.log_event("Simulation environment initialized.")
test_stand.log_event("Virtual sensors initialized.")
test_stand.log_event("Telemetry system online.")
test_stand.log_event("Safety monitoring initialized.")

print("Simulation core initialized.")
print("System Status:", test_stand.status)
print("Virtual Sensors:", len(test_stand.sensors))
print("Telemetry Storage: READY")
print("Event Logging: READY")

def generate_telemetry(t):
    phase = "IDLE"

    if t < 10:
        phase = "INITIALIZATION"
        temperature = 25 + np.random.normal(0, 0.3)

    elif t < 20:
        phase = "STARTUP SIMULATION"
        temperature = 25 + (t - 10) * 4

    elif t < 45:
        phase = "STEADY STATE SIMULATION"
        temperature = 65 + np.random.normal(0, 1)

    else:
        phase = "SHUTDOWN SIMULATION"
        temperature = max(25, 65 - (t - 45) * 2)

    pressure = max(0, 50 * np.sin(np.pi * min(t, 60) / 60))
    vibration = abs(np.random.normal(0.5, 0.15))
    fuel_flow = max(0, 20 * np.sin(np.pi * min(t, 60) / 60))
    oxidizer_flow = max(0, 22 * np.sin(np.pi * min(t, 60) / 60))
    tank_pressure = max(0, 100 - t * 0.3)

    readings = {
        "Chamber Pressure": pressure,
        "Engine Temperature": temperature,
        "Fuel Flow": fuel_flow,
        "Oxidizer Flow": oxidizer_flow,
        "Vibration": vibration,
        "Tank Pressure": tank_pressure
    }

    return phase, readings


def run_test_sequence(duration=60):

    if test_stand.status == "RUNNING":
        print("A simulation is already running.")
        return

    if duration <= 0 or duration > 120:
        print("Duration must be between 1 and 120 seconds.")
        return

    test_stand.reset()
    test_stand.status = "RUNNING"
    test_stand.log_event("Virtual test sequence started.")

    print("Simulation engine ready.")
    print("Test duration:", duration, "seconds")

    for t in range(duration):

        phase, readings = generate_telemetry(t)

        test_stand.phase = phase
        test_stand.elapsed_time = t
        test_stand.update_sensors(readings)
        test_stand.record_telemetry()

        time.sleep(0.05)

    test_stand.status = "COMPLETE"
    test_stand.phase = "COMPLETE"
    test_stand.log_event("Virtual test sequence completed.")

    print("Simulation completed successfully.")
    print("Telemetry samples:", len(test_stand.telemetry))
    print("System Status:", test_stand.status)


print("Simulation engine loaded successfully.")

def display_dashboard():

    if not test_stand.telemetry:
        print("No telemetry data available. Run a simulation first.")
        return

    df = test_stand.get_telemetry()

    latest = df.iloc[-1]

    print("ROCKET TEST STAND | TELEMETRY DASHBOARD")
    print("=" * 55)
    print("SYSTEM STATUS:", test_stand.status)
    print("TEST PHASE:", test_stand.phase)
    print("SIMULATION TIME:", latest["Time"], "seconds")
    print("=" * 55)

    print("\nLIVE SENSOR READINGS")

    for sensor in test_stand.sensors:
        print(f"{sensor}: {latest[sensor]:.2f}")

    fig, axes = plt.subplots(2, 2, figsize=(14, 8))

    axes[0, 0].plot(
        df["Time"], df["Chamber Pressure"], linewidth=2
    )
    axes[0, 0].set_title("Chamber Pressure")
    axes[0, 0].set_ylabel("Simulated Units")

    axes[0, 1].plot(
        df["Time"], df["Engine Temperature"], linewidth=2
    )
    axes[0, 1].set_title("Engine Temperature")
    axes[0, 1].set_ylabel("Degrees C")

    axes[1, 0].plot(
        df["Time"], df["Fuel Flow"], label="Fuel"
    )
    axes[1, 0].plot(
        df["Time"], df["Oxidizer Flow"], label="Oxidizer"
    )
    axes[1, 0].set_title("Simulated Flow Measurements")
    axes[1, 0].legend()

    axes[1, 1].plot(
        df["Time"], df["Vibration"], linewidth=2
    )
    axes[1, 1].set_title("Vibration Monitoring")
    axes[1, 1].set_ylabel("Simulated Units")

    for ax in axes.flat:
        ax.set_xlabel("Simulation Time (seconds)")
        ax.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()


print("Telemetry dashboard loaded successfully.")

def start_simulation(button):
    clear_output(wait=True)
    run_test_sequence(duration=60)
    display_controls()


def show_dashboard(button):
    clear_output(wait=True)
    display_dashboard()
    display_controls()


def reset_simulation(button):
    clear_output(wait=True)
    test_stand.reset()
    display_controls()


def show_event_log(button):
    clear_output(wait=True)

    print("TEST STAND EVENT LOG")

    if test_stand.events:
        display(test_stand.get_event_log())
    else:
        print("No events recorded.")

    display_controls()


start_button = widgets.Button(
    description="Start Simulation",
    button_style="success",
    icon="play"
)

dashboard_button = widgets.Button(
    description="View Dashboard",
    button_style="info",
    icon="bar-chart"
)

reset_button = widgets.Button(
    description="Reset System",
    button_style="warning",
    icon="refresh"
)

log_button = widgets.Button(
    description="Event Log",
    icon="list"
)

start_button.on_click(start_simulation)
dashboard_button.on_click(show_dashboard)
reset_button.on_click(reset_simulation)
log_button.on_click(show_event_log)


def display_controls():

    print("ROCKET TEST STAND CONTROL PANEL")
    print("System Status:", test_stand.status)
    print("Current Phase:", test_stand.phase)

    display(widgets.HBox([
        start_button,
        dashboard_button
    ]))

    display(widgets.HBox([
        reset_button,
        log_button
    ]))


display_controls()

from google.colab import files

def export_telemetry(button):

    if not test_stand.telemetry:
        print("No telemetry available. Run a simulation first.")
        return

    df = test_stand.get_telemetry()

    filename = "rocket_test_stand_telemetry.csv"

    df.to_csv(filename, index=False)

    print("Telemetry export completed.")
    print("Records exported:", len(df))
    print("Filename:", filename)

    files.download(filename)


def export_events(button):

    if not test_stand.events:
        print("No event records available.")
        return

    df = test_stand.get_event_log()

    filename = "rocket_test_stand_events.csv"

    df.to_csv(filename, index=False)

    print("Event log export completed.")

    files.download(filename)


telemetry_export_button = widgets.Button(
    description="Export Telemetry CSV",
    button_style="success",
    icon="download"
)

event_export_button = widgets.Button(
    description="Export Event Log",
    button_style="info",
    icon="file"
)

telemetry_export_button.on_click(export_telemetry)
event_export_button.on_click(export_events)

print("DATA ACQUISITION AND EXPORT")

display(widgets.HBox([
    telemetry_export_button,
    event_export_button
]))

def evaluate_safety(readings):

    alerts = []

    if readings["Engine Temperature"] > 80:
        alerts.append("WARNING: Temperature threshold exceeded.")

    if readings["Vibration"] > 1.2:
        alerts.append("WARNING: Elevated vibration detected.")

    if readings["Tank Pressure"] < 20:
        alerts.append("WARNING: Low simulated tank pressure.")

    if readings["Chamber Pressure"] < 0:
        alerts.append("WARNING: Invalid pressure reading.")

    return alerts


def run_safety_check(button):

    if not test_stand.telemetry:
        print("No telemetry available for safety evaluation.")
        return

    latest = test_stand.telemetry[-1]

    alerts = evaluate_safety(latest)

    print("SAFETY MONITORING REPORT")
    print("System Status:", test_stand.status)

    if alerts:
        for alert in alerts:
            print(alert)
            test_stand.log_event(alert)

    else:
        print("No simulated safety warnings detected.")

    print("Safety evaluation completed.")


safety_button = widgets.Button(
    description="Run Safety Check",
    button_style="danger",
    icon="shield"
)

safety_button.on_click(run_safety_check)

display(safety_button)

print("Safety monitoring module loaded.")

def integrated_dashboard():

    if not test_stand.telemetry:
        print("No simulation data available.")
        return

    df = test_stand.get_telemetry()
    latest = df.iloc[-1]

    alerts = evaluate_safety(latest)

    print("=" * 65)
    print("             ROCKET ENGINE TEST OPERATIONS")
    print("=" * 65)

    print(f" SYSTEM STATUS: {test_stand.status}")
    print(f" TEST PHASE: {test_stand.phase}")
    print(f" ELAPSED TIME: {latest['Time']} seconds")
    print(f" TELEMETRY RECORDS: {len(df)}")

    print("\n SENSOR MONITORING")
    print("-" * 65)

    for sensor in test_stand.sensors:
        print(f"{sensor:<25} {latest[sensor]:>12.2f}")

    print("\n SAFETY MONITORING")
    print("-" * 65)

    if alerts:
        for alert in alerts:
            print(alert)
    else:
        print("ALL SIMULATED SENSOR CHECKS NORMAL")

    fig, axes = plt.subplots(3, 2, figsize=(15, 11))

    chart_configs = [
        ("Chamber Pressure", "Chamber Pressure"),
        ("Engine Temperature", "Engine Temperature"),
        ("Fuel Flow", "Fuel Flow"),
        ("Oxidizer Flow", "Oxidizer Flow"),
        ("Vibration", "Vibration"),
        ("Tank Pressure", "Tank Pressure")
    ]

    for ax, (column, title) in zip(axes.flat, chart_configs):

        ax.plot(
            df["Time"],
            df[column],
            linewidth=2
        )

        ax.set_title(title)
        ax.set_xlabel("Simulation Time (seconds)")
        ax.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

    print("\n RECENT EVENT HISTORY")
    print("-" * 65)

    if test_stand.events:
        display(test_stand.get_event_log().tail(10))

    print("\n DASHBOARD UPDATE COMPLETE")


dashboard_button_full = widgets.Button(
    description="Open Engineering Dashboard",
    button_style="primary",
    icon="dashboard"
)


def open_full_dashboard(button):
    clear_output(wait=True)
    integrated_dashboard()
    display(dashboard_button_full)


dashboard_button_full.on_click(open_full_dashboard)

print("Integrated engineering dashboard loaded.")
display(dashboard_button_full)

run_test_sequence(duration=60)

def display_operations_interface():

    if not test_stand.telemetry:
        print("Run a simulation before opening the operations interface.")
        return

    df = test_stand.get_telemetry()
    latest = df.iloc[-1]

    alerts = evaluate_safety(latest)

    status_color = {
        "READY": "#16a34a",
        "RUNNING": "#2563eb",
        "COMPLETE": "#16a34a",
        "OFFLINE": "#64748b"
    }.get(test_stand.status, "#d97706")

    alert_color = "#dc2626" if alerts else "#16a34a"
    alert_text = "WARNINGS DETECTED" if alerts else "NO ACTIVE WARNINGS"

    header = widgets.HTML(value=f"""
    <div style="background:#172554;color:white;padding:22px;
                border-radius:10px;margin-bottom:12px;">
        <h2 style="margin:0;">ENGINE TEST OPERATIONS</h2>
        <p style="margin:6px 0 0;color:#bfdbfe;">
            Virtual Test Stand | Monitoring and Data Acquisition
        </p>
        <div style="margin-top:15px;">
            <span style="background:{status_color};padding:7px 14px;
                         border-radius:5px;color:white;">
                {test_stand.status}
            </span>
            <span style="margin-left:12px;color:#bfdbfe;">
                Phase: {test_stand.phase}
            </span>
        </div>
    </div>
    """)

    sensor_cards = []

    for sensor in test_stand.sensors:

        value = latest[sensor]

        sensor_cards.append(widgets.HTML(value=f"""
        <div style="background:#f1f5f9;border:1px solid #cbd5e1;
                    border-radius:8px;padding:15px;min-width:175px;">
            <div style="font-size:12px;color:#64748b;">
                {sensor.upper()}
            </div>
            <div style="font-size:25px;font-weight:bold;
                        color:#0f172a;margin-top:8px;">
                {value:.2f}
            </div>
            <div style="font-size:11px;color:#64748b;">
                SIMULATED READING
            </div>
        </div>
        """))

    sensor_grid = widgets.GridBox(
        sensor_cards,
        layout=widgets.Layout(
            grid_template_columns="repeat(3, 1fr)",
            grid_gap="10px",
            width="100%"
        )
    )

    alert_panel = widgets.HTML(value=f"""
    <div style="background:#f8fafc;border-left:5px solid {alert_color};
                padding:16px;border-radius:6px;margin-top:15px;">
        <h3 style="margin:0;color:#0f172a;">SAFETY MONITORING</h3>
        <p style="color:{alert_color};font-weight:bold;">
            {alert_text}
        </p>
        <p style="color:#475569;font-size:13px;">
            {"<br>".join(alerts) if alerts else
             "All configured virtual sensor checks passed."}
        </p>
    </div>
    """)

    progress = widgets.IntProgress(
        value=min(100, int(len(df) / 60 * 100)),
        min=0,
        max=100,
        description="Test Progress:",
        bar_style="success",
        layout=widgets.Layout(width="100%")
    )

    print("SYSTEM OVERVIEW")

    display(header)

    print("VIRTUAL SENSOR TELEMETRY")

    display(sensor_grid)

    print("TEST SEQUENCE")

    display(progress)

    display(alert_panel)

    print("TEST STAND SCHEMATIC")

    fig, ax = plt.subplots(figsize=(12, 4))

    components = [
        (0.05, 0.4, "FUEL TANK"),
        (0.27, 0.4, "OXIDIZER TANK"),
        (0.50, 0.4, "TEST CHAMBER"),
        (0.76, 0.4, "DATA SYSTEM")
    ]

    for x, y, label in components:

        rect = plt.Rectangle(
            (x, y),
            0.17,
            0.25,
            facecolor="#dbeafe",
            edgecolor="#1e40af",
            linewidth=2
        )

        ax.add_patch(rect)

        ax.text(
            x + 0.085,
            y + 0.125,
            label,
            ha="center",
            va="center",
            fontsize=9,
            fontweight="bold"
        )

    for start, end in [(0.22, 0.50), (0.44, 0.50), (0.67, 0.76)]:

        ax.annotate(
            "",
            xy=(end, 0.525),
            xytext=(start, 0.525),
            arrowprops=dict(arrowstyle="->", linewidth=2)
        )

    ax.text(
        0.5, 0.85,
        "VIRTUAL TEST STAND ARCHITECTURE",
        ha="center",
        fontsize=13,
        fontweight="bold"
    )

    ax.text(
        0.5, 0.15,
        "Schematic representation | Not to scale",
        ha="center",
        fontsize=9,
        color="gray"
    )

    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.axis("off")

    plt.show()

    print("DATA ACQUISITION SUMMARY")
    print("Total Records:", len(df))
    print("Latest Simulation Time:", latest["Time"], "seconds")
    print("Event Records:", len(test_stand.events))


display_operations_interface()

def inject_sensor_fault(sensor_name, fault_type):

    if not test_stand.telemetry:
        print("No telemetry available.")
        return

    df = test_stand.get_telemetry().copy()

    if len(df) < 10:
        print("At least 10 telemetry records are required.")
        return

    original = df[sensor_name].copy()

    if fault_type == "Missing Data":
        df.loc[df.index[-10:], sensor_name] = np.nan

    elif fault_type == "Stuck Sensor":
        df.loc[df.index[-10:], sensor_name] = original.iloc[-11]

    elif fault_type == "Abnormal Reading":
        median = original.median()
        spread = original.std()

        df.loc[df.index[-1], sensor_name] = median + 20 * max(spread, 1)

    return df


def diagnose_sensor(df, sensor_name):

    readings = df[sensor_name]
    recent = readings.tail(10)

    issues = []

    if recent.isna().any():
        issues.append("Missing sensor data detected.")

    valid = recent.dropna()

    if len(valid) == 10 and valid.nunique() == 1:
        issues.append("Sensor output appears stuck.")

    if len(readings.dropna()) >= 10 and not readings.isna().all():

        historical = readings.dropna()

        median = historical.median()
        mad = np.median(np.abs(historical - median))

        if mad > 0 and abs(historical.iloc[-1] - median) > 8 * mad:
            issues.append("Abnormal sensor deviation detected.")

    if issues:
        return issues

    return ["No simulated sensor faults detected."]


def run_diagnostics(button):

    sensor = sensor_selector.value
    fault = fault_selector.value

    output = inject_sensor_fault(sensor, fault)

    with diagnostic_output:
        clear_output(wait=True)

        print("SENSOR DIAGNOSTIC REPORT")
        print("Sensor:", sensor)
        print("Injected Condition:", fault)
        print()

        results = diagnose_sensor(output, sensor)

        for result in results:
            print("[ALERT]", result)

        print()
        print("Diagnostic analysis completed.")
        print("Original telemetry records remain unchanged.")


sensor_selector = widgets.Dropdown(
    options=list(test_stand.sensors.keys()),
    description="Sensor:",
    layout=widgets.Layout(width="400px")
)

fault_selector = widgets.Dropdown(
    options=[
        "Missing Data",
        "Stuck Sensor",
        "Abnormal Reading"
    ],
    description="Fault:",
    layout=widgets.Layout(width="400px")
)

diagnostic_button = widgets.Button(
    description="Run Diagnostics",
    button_style="danger",
    icon="search"
)

diagnostic_output = widgets.Output()

diagnostic_button.on_click(run_diagnostics)

display(widgets.VBox([
    sensor_selector,
    fault_selector,
    diagnostic_button,
    diagnostic_output
]))

print("Sensor diagnostics module initialized.")

import sqlite3
import uuid

DATABASE = "rocket_test_stand.db"

def save_to_database():

    if not test_stand.telemetry:
        print("No telemetry data available to save.")
        return

    run_id = str(uuid.uuid4())

    connection = sqlite3.connect(DATABASE)
    cursor = connection.cursor()

    cursor.execute("""
        CREATE TABLE IF NOT EXISTS telemetry (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            run_id TEXT,
            time REAL,
            timestamp TEXT,
            chamber_pressure REAL,
            engine_temperature REAL,
            fuel_flow REAL,
            oxidizer_flow REAL,
            vibration REAL,
            tank_pressure REAL
        )
    """)

    cursor.execute("""
        CREATE TABLE IF NOT EXISTS events (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            run_id TEXT,
            timestamp TEXT,
            event TEXT
        )
    """)

    telemetry_records = []

    for record in test_stand.telemetry:
        telemetry_records.append((
            run_id,
            record["Time"],
            record["Timestamp"],
            record["Chamber Pressure"],
            record["Engine Temperature"],
            record["Fuel Flow"],
            record["Oxidizer Flow"],
            record["Vibration"],
            record["Tank Pressure"]
        ))

    cursor.executemany("""
        INSERT INTO telemetry (
            run_id, time, timestamp, chamber_pressure,
            engine_temperature, fuel_flow, oxidizer_flow,
            vibration, tank_pressure
        )
        VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?)
    """, telemetry_records)

    event_records = [
        (run_id, event["Timestamp"], event["Event"])
        for event in test_stand.events
    ]

    cursor.executemany("""
        INSERT INTO events (run_id, timestamp, event)
        VALUES (?, ?, ?)
    """, event_records)

    connection.commit()
    connection.close()

    print("DATABASE SAVE COMPLETED")
    print("Run ID:", run_id)
    print("Telemetry records saved:", len(telemetry_records))
    print("Event records saved:", len(event_records))


def query_database():

    connection = sqlite3.connect(DATABASE)

    telemetry_df = pd.read_sql_query(
        "SELECT * FROM telemetry ORDER BY id",
        connection
    )

    events_df = pd.read_sql_query(
        "SELECT * FROM events ORDER BY id",
        connection
    )

    connection.close()

    print("DATABASE SUMMARY")
    print("Telemetry Records:", len(telemetry_df))
    print("Event Records:", len(events_df))

    return telemetry_df, events_df


database_button = widgets.Button(
    description="Save to Database",
    button_style="success",
    icon="database"
)

database_button.on_click(lambda button: save_to_database())

display(database_button)

print("SQLite database module initialized.")

def display_database_records(button):

    telemetry_df, events_df = query_database()

    clear_output(wait=True)

    print("DATABASE TELEMETRY RECORDS")
    display(telemetry_df.tail(10))

    print("\nDATABASE EVENT HISTORY")
    display(events_df.tail(10))

    print("\nDATABASE STATISTICS")

    if not telemetry_df.empty:

        print("Total Test Runs:", telemetry_df["run_id"].nunique())

        print(
            "Average Engine Temperature:",
            round(telemetry_df["engine_temperature"].mean(), 2)
        )

        print(
            "Maximum Chamber Pressure:",
            round(telemetry_df["chamber_pressure"].max(), 2)
        )

        print(
            "Average Vibration:",
            round(telemetry_df["vibration"].mean(), 2)
        )


database_view_button = widgets.Button(
    description="View Database Records",
    button_style="info",
    icon="table"
)

database_view_button.on_click(display_database_records)

display(database_view_button)

print("Database analysis module loaded.")

!pip install -q reportlab

from reportlab.lib import colors
from reportlab.lib.pagesizes import letter
from reportlab.platypus import (
    SimpleDocTemplate,
    Paragraph,
    Spacer,
    Table,
    TableStyle
)
from reportlab.lib.styles import getSampleStyleSheet
from reportlab.lib.units import inch


def generate_test_report(button):

    if not test_stand.telemetry:
        print("Run a simulation before generating a report.")
        return

    df = test_stand.get_telemetry()
    latest = df.iloc[-1]
    alerts = evaluate_safety(latest)

    filename = "rocket_test_stand_report.pdf"

    document = SimpleDocTemplate(
        filename,
        pagesize=letter
    )

    styles = getSampleStyleSheet()
    content = []

    content.append(
        Paragraph("ROCKET TEST STAND", styles["Title"])
    )

    content.append(
        Paragraph("Virtual Engine Test Report", styles["Heading2"])
    )

    content.append(Spacer(1, 20))

    summary = [
        ["Report Information", "Result"],
        ["System Status", test_stand.status],
        ["Test Phase", test_stand.phase],
        ["Simulation Duration", f"{latest['Time']} seconds"],
        ["Telemetry Records", str(len(df))],
        ["Report Generated", datetime.now().strftime("%Y-%m-%d %H:%M:%S")]
    ]

    table = Table(summary, colWidths=[2.5 * inch, 3.5 * inch])

    table.setStyle(TableStyle([
        ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#172554")),
        ("TEXTCOLOR", (0, 0), (-1, 0), colors.white),
        ("GRID", (0, 0), (-1, -1), 0.5, colors.grey),
        ("PADDING", (0, 0), (-1, -1), 8)
    ]))

    content.append(table)
    content.append(Spacer(1, 20))

    content.append(
        Paragraph("Sensor Performance Summary", styles["Heading2"])
    )

    sensor_data = [
        ["Sensor", "Average", "Maximum", "Minimum"]
    ]

    for sensor in test_stand.sensors:

        sensor_data.append([
            sensor,
            f"{df[sensor].mean():.2f}",
            f"{df[sensor].max():.2f}",
            f"{df[sensor].min():.2f}"
        ])

    sensor_table = Table(
        sensor_data,
        colWidths=[2.0 * inch, 1.2 * inch, 1.2 * inch, 1.2 * inch]
    )

    sensor_table.setStyle(TableStyle([
        ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#1e40af")),
        ("TEXTCOLOR", (0, 0), (-1, 0), colors.white),
        ("GRID", (0, 0), (-1, -1), 0.5, colors.grey),
        ("PADDING", (0, 0), (-1, -1), 7)
    ]))

    content.append(sensor_table)
    content.append(Spacer(1, 20))

    content.append(
        Paragraph("Safety Monitoring Results", styles["Heading2"])
    )

    if alerts:
        for alert in alerts:
            content.append(Paragraph(alert, styles["BodyText"]))
    else:
        content.append(
            Paragraph(
                "No simulated safety warnings detected.",
                styles["BodyText"]
            )
        )

    content.append(Spacer(1, 20))

    content.append(
        Paragraph("Event History", styles["Heading2"])
    )

    for event in test_stand.events[-10:]:

        content.append(
            Paragraph(
                f"{event['Timestamp']} - {event['Event']}",
                styles["BodyText"]
            )
        )

    content.append(Spacer(1, 20))

    content.append(
        Paragraph(
            "This report contains simulated data generated by a "
            "virtual test stand environment. It does not represent "
            "actual engine testing or hardware measurements.",
            styles["Italic"]
        )
    )

    document.build(content)

    print("TEST REPORT GENERATED")
    print("Filename:", filename)

    from google.colab import files
    files.download(filename)


report_button = widgets.Button(
    description="Generate PDF Test Report",
    button_style="success",
    icon="file-pdf"
)

report_button.on_click(generate_test_report)

display(report_button)

print("Automated reporting module initialized.")

live_history = []

live_play = widgets.Play(
    value=0,
    min=0,
    max=59,
    step=1,
    interval=500,
    description="Run"
)

live_slider = widgets.IntSlider(
    value=0,
    min=0,
    max=59,
    step=1,
    description="Test Time:",
    continuous_update=True
)

widgets.jslink((live_play, "value"), (live_slider, "value"))

live_output = widgets.Output()


def update_live_monitor(change):

    t = change["new"]

    if t == 0:
        live_history.clear()

    phase, readings = generate_telemetry(t)

    live_history.append({
        "Time": t,
        **readings
    })

    with live_output:
        clear_output(wait=True)

        print("LIVE ENGINE TEST MONITORING")
        print("=" * 50)
        print("SYSTEM STATUS:", "COMPLETE" if t == 59 else "RUNNING")
        print("TEST PHASE:", phase)
        print("SIMULATION TIME:", t, "seconds")

        sensor_cards = []

        for sensor, value in readings.items():

            sensor_cards.append(widgets.HTML(value=f"""
            <div style="background:#f1f5f9;padding:14px;
                        border:1px solid #cbd5e1;border-radius:8px;">
                <div style="font-size:11px;color:#64748b;">
                    {sensor.upper()}
                </div>
                <div style="font-size:23px;font-weight:bold;
                            color:#0f172a;">
                    {value:.2f}
                </div>
            </div>
            """))

        display(widgets.GridBox(
            sensor_cards,
            layout=widgets.Layout(
                grid_template_columns="repeat(3, 1fr)",
                grid_gap="10px",
                width="100%"
            )
        ))

        history_df = pd.DataFrame(live_history)

        fig, axes = plt.subplots(2, 2, figsize=(12, 7))

        chart_settings = [
            ("Chamber Pressure", "Chamber Pressure"),
            ("Engine Temperature", "Engine Temperature"),
            ("Fuel Flow", "Fuel Flow"),
            ("Vibration", "Vibration")
        ]

        for ax, (column, title) in zip(axes.flat, chart_settings):

            ax.plot(
                history_df["Time"],
                history_df[column],
                linewidth=2
            )

            ax.set_title(title)
            ax.set_xlabel("Simulation Time")
            ax.grid(True, alpha=0.3)

        plt.tight_layout()
        plt.show()

        print("Telemetry Samples:", len(live_history))


live_slider.observe(update_live_monitor, names="value")

display(widgets.VBox([
    widgets.HTML("<h3>VIRTUAL ENGINE TEST STAND</h3>"),
    live_play,
    live_slider,
    live_output
]))

update_live_monitor({"new": 0})

print("Live monitoring interface initialized.")

def detect_anomalies(df, threshold=3.5):

    sensor_columns = [
        "Chamber Pressure",
        "Engine Temperature",
        "Fuel Flow",
        "Oxidizer Flow",
        "Vibration",
        "Tank Pressure"
    ]

    anomalies = []

    for sensor in sensor_columns:

        readings = df[sensor].dropna()

        if len(readings) < 10:
            continue

        median = readings.median()

        mad = np.median(
            np.abs(readings - median)
        )

        if mad > 0:
            scores = 0.6745 * (readings - median) / mad

        else:
            std = readings.std()

            if std > 0:
                scores = (readings - median) / std
            else:
                scores = pd.Series(
                    np.where(readings == median, 0, np.inf),
                    index=readings.index
                )

        unusual = readings[abs(scores) > threshold]

        for index, value in unusual.items():

            anomalies.append({
                "Time": df.loc[index, "Time"],
                "Sensor": sensor,
                "Reading": round(value, 2),
                "Historical Median": round(median, 2),
                "Anomaly Score": round(scores.loc[index], 2)
            })

    return pd.DataFrame(anomalies)


def run_anomaly_analysis(button):

    with anomaly_output:
        clear_output(wait=True)

        if not test_stand.telemetry:
            print("No telemetry available for analysis.")
            return

        df = test_stand.get_telemetry()

        print("AUTOMATED ANOMALY DETECTION")
        print("=" * 50)

        results = detect_anomalies(df)

        if results.empty:

            print("No significant anomalies detected.")
            print("All analyzed readings are within the configured threshold.")

        else:

            print("ANOMALIES DETECTED:", len(results))
            display(results)

        print("\nSENSORS ANALYZED:", 6)
        print("ANALYSIS COMPLETE")
        print("Original telemetry records remain unchanged.")


anomaly_button = widgets.Button(
    description="Run Anomaly Detection",
    button_style="danger",
    icon="search"
)

anomaly_output = widgets.Output()

anomaly_button.on_click(run_anomaly_analysis)

display(widgets.VBox([
    anomaly_button,
    anomaly_output
]))

print("Automated anomaly detection module initialized.")

def create_comparison_run(button):

    with comparison_output:
        clear_output(wait=True)

        print("GENERATING NEW TEST RUN")

        run_test_sequence(duration=60)

        save_to_database()

        print("\nNew test run saved successfully.")
        print("You can now compare your saved runs.")


def compare_test_runs(button):

    with comparison_output:
        clear_output(wait=True)

        connection = sqlite3.connect(DATABASE)

        df = pd.read_sql_query(
            "SELECT * FROM telemetry ORDER BY id",
            connection
        )

        connection.close()

        if df["run_id"].nunique() < 2:
            print("At least two saved test runs are required.")
            return

        comparison = df.groupby("run_id").agg(
            Samples=("time", "count"),
            Average_Temperature=("engine_temperature", "mean"),
            Maximum_Temperature=("engine_temperature", "max"),
            Maximum_Chamber_Pressure=("chamber_pressure", "max"),
            Average_Vibration=("vibration", "mean"),
            Minimum_Tank_Pressure=("tank_pressure", "min")
        ).reset_index()

        comparison["Run"] = [
            f"Run {i + 1}"
            for i in range(len(comparison))
        ]

        comparison = comparison.round(2)

        print("TEST RUN PERFORMANCE COMPARISON")
        print("=" * 60)

        print("Total Saved Runs:", len(comparison))

        display(comparison[[
            "Run",
            "Samples",
            "Average_Temperature",
            "Maximum_Temperature",
            "Maximum_Chamber_Pressure",
            "Average_Vibration",
            "Minimum_Tank_Pressure"
        ]])

        fig, axes = plt.subplots(1, 2, figsize=(14, 5))

        axes[0].bar(
            comparison["Run"],
            comparison["Average_Temperature"]
        )

        axes[0].set_title("Average Engine Temperature")
        axes[0].set_ylabel("Degrees C")
        axes[0].grid(axis="y", alpha=0.3)

        axes[1].bar(
            comparison["Run"],
            comparison["Maximum_Chamber_Pressure"]
        )

        axes[1].set_title("Maximum Chamber Pressure")
        axes[1].set_ylabel("Simulated Units")
        axes[1].grid(axis="y", alpha=0.3)

        plt.tight_layout()
        plt.show()

        print("TEST RUN COMPARISON COMPLETED")


new_run_button = widgets.Button(
    description="Generate and Save New Test Run",
    button_style="success",
    icon="play"
)

compare_runs_button = widgets.Button(
    description="Compare Saved Test Runs",
    button_style="info",
    icon="bar-chart"
)

comparison_output = widgets.Output()

new_run_button.on_click(create_comparison_run)
compare_runs_button.on_click(compare_test_runs)

display(widgets.VBox([
    new_run_button,
    compare_runs_button,
    comparison_output
]))

print("Test-run comparison module initialized.")

def display_unified_dashboard():

    if not test_stand.telemetry:
        print("Run a simulation before opening the dashboard.")
        return

    df = test_stand.get_telemetry()
    latest = df.iloc[-1]

    alerts = evaluate_safety(latest)
    anomalies = detect_anomalies(df)

    connection = sqlite3.connect(DATABASE)

    database_runs = pd.read_sql_query(
        "SELECT COUNT(DISTINCT run_id) AS total FROM telemetry",
        connection
    ).iloc[0]["total"]

    database_records = pd.read_sql_query(
        "SELECT COUNT(*) AS total FROM telemetry",
        connection
    ).iloc[0]["total"]

    connection.close()

    status_color = (
        "#16a34a" if test_stand.status == "COMPLETE"
        else "#2563eb"
    )

    warning_color = "#dc2626" if alerts or not anomalies.empty else "#16a34a"

    header = widgets.HTML(value=f"""
    <div style="background:#172554;color:white;padding:24px;
                border-radius:10px;">
        <h1 style="margin:0;">ENGINEERING OPERATIONS CENTER</h1>
        <p style="color:#bfdbfe;margin-bottom:15px;">
            Virtual Rocket Test Stand | System Monitoring
        </p>
        <span style="background:{status_color};padding:8px 14px;
                     border-radius:5px;font-weight:bold;">
            {test_stand.status}
        </span>
        <span style="margin-left:15px;color:#bfdbfe;">
            Current Phase: {test_stand.phase}
        </span>
    </div>
    """)

    metrics = [
        ("TELEMETRY RECORDS", len(df)),
        ("SAVED TEST RUNS", database_runs),
        ("SAFETY ALERTS", len(alerts)),
        ("ANOMALIES", len(anomalies))
    ]

    metric_cards = []

    for label, value in metrics:

        metric_cards.append(widgets.HTML(value=f"""
        <div style="background:#f1f5f9;padding:18px;
                    border-radius:8px;border:1px solid #cbd5e1;">
            <div style="font-size:11px;color:#64748b;">
                {label}
            </div>
            <div style="font-size:27px;font-weight:bold;
                        color:#0f172a;margin-top:8px;">
                {value}
            </div>
        </div>
        """))

    metric_grid = widgets.GridBox(
        metric_cards,
        layout=widgets.Layout(
            grid_template_columns="repeat(4, 1fr)",
            grid_gap="10px",
            width="100%"
        )
    )

    sensor_data = []

    for sensor in test_stand.sensors:

        sensor_data.append([
            sensor,
            f"{latest[sensor]:.2f}"
        ])

    sensor_table = widgets.HTML(value=f"""
    <div style="background:#f8fafc;padding:18px;
                border:1px solid #cbd5e1;border-radius:8px;">
        <h3 style="color:#0f172a;">LIVE SENSOR STATUS</h3>
        <table style="width:100%;border-collapse:collapse;">
            {''.join(
                f'<tr style="border-bottom:1px solid #e2e8f0;">'
                f'<td style="padding:9px;color:#475569;">{name}</td>'
                f'<td style="padding:9px;text-align:right;'
                f'font-weight:bold;color:#0f172a;">{value}</td>'
                f'</tr>'
                for name, value in sensor_data
            )}
        </table>
    </div>
    """)

    safety_message = (
        "Simulated safety warnings detected."
        if alerts else
        "No configured safety warnings detected."
    )

    anomaly_message = (
        f"{len(anomalies)} unusual readings identified."
        if not anomalies.empty else
        "No significant statistical anomalies detected."
    )

    monitoring_panel = widgets.HTML(value=f"""
    <div style="background:#f8fafc;padding:18px;
                border-left:5px solid {warning_color};
                border-radius:8px;">
        <h3 style="color:#0f172a;">MONITORING AND DIAGNOSTICS</h3>
        <p style="color:{warning_color};font-weight:bold;">
            {safety_message}
        </p>
        <p style="color:#475569;">{anomaly_message}</p>
        <p style="color:#64748b;font-size:12px;">
            Statistical anomaly detection uses historical sensor readings.
        </p>
    </div>
    """)

    fig, axes = plt.subplots(2, 2, figsize=(13, 8))

    chart_settings = [
        ("Chamber Pressure", "Chamber Pressure"),
        ("Engine Temperature", "Engine Temperature"),
        ("Fuel Flow", "Fuel Flow"),
        ("Vibration", "Vibration")
    ]

    for ax, (column, title) in zip(axes.flat, chart_settings):

        ax.plot(
            df["Time"],
            df[column],
            linewidth=2
        )

        ax.set_title(title)
        ax.set_xlabel("Simulation Time (seconds)")
        ax.grid(True, alpha=0.3)

    plt.tight_layout()

    print("ENGINEERING OPERATIONS CENTER")

    display(header)

    print("\nSYSTEM PERFORMANCE")

    display(metric_grid)

    print("\nSENSOR TELEMETRY")

    display(sensor_table)

    print("\nSAFETY AND DIAGNOSTICS")

    display(monitoring_panel)

    print("\nTELEMETRY VISUALIZATION")

    plt.show()

    print("\nDATABASE INFORMATION")
    print("Stored Telemetry Records:", database_records)
    print("Stored Test Runs:", database_runs)
    print("Recorded Events:", len(test_stand.events))

    print("\nOPERATIONS DASHBOARD COMPLETE")


unified_dashboard_button = widgets.Button(
    description="Launch Engineering Operations Center",
    button_style="primary",
    icon="dashboard"
)


def launch_unified_dashboard(button):

    clear_output(wait=True)
    display_unified_dashboard()
    display(unified_dashboard_button)


unified_dashboard_button.on_click(launch_unified_dashboard)

display(unified_dashboard_button)

print("Unified engineering dashboard initialized.")

project_readme = """
# Virtual Rocket Engine Test Stand Simulator

## Project Overview

The Virtual Rocket Engine Test Stand Simulator is a Python-based
engineering software project designed to simulate an engine test
environment, collect virtual sensor telemetry, monitor system
conditions, identify anomalies, and store historical test data.

The project demonstrates how software engineering, data analysis,
database management, and interactive visualization can be combined
into an integrated monitoring application.

All sensor readings and test sequences are simulated.
The application does not control physical hardware.

## Technologies Used

- Python
- NumPy
- Pandas
- Matplotlib
- SQLite
- IPyWidgets
- ReportLab
- Google Colab

## System Architecture

The application is organized into several interconnected modules.

1. Simulation Engine
   Generates virtual sensor readings and manages test sequences.

2. Telemetry Acquisition
   Collects and organizes simulated sensor measurements.

3. Monitoring Dashboard
   Displays sensor readings, system status, and performance graphs.

4. Safety Monitoring
   Evaluates configured sensor thresholds and generates warnings.

5. Sensor Diagnostics
   Simulates sensor faults, including missing data, stuck readings,
   and abnormal measurements.

6. Anomaly Detection
   Uses statistical analysis and median absolute deviation to
   identify unusual telemetry readings.

7. Database Management
   Uses SQLite to store telemetry records and event history.
   Each saved snapshot receives a unique run identifier.

8. Performance Analysis
   Compares historical test runs and calculates sensor statistics.

9. Automated Reporting
   Generates downloadable PDF reports containing test information,
   sensor statistics, safety results, and event history.

## Virtual Sensor Measurements

The simulator monitors six virtual sensors:

- Chamber Pressure
- Engine Temperature
- Fuel Flow
- Oxidizer Flow
- Vibration
- Tank Pressure

## Major Features

- Virtual test sequence execution
- Interactive operator controls
- Live telemetry visualization
- Sensor fault injection
- Automated safety evaluation
- Statistical anomaly detection
- SQLite database integration
- Historical test-run comparison
- CSV data export
- Automated PDF report generation
- Unified engineering operations dashboard

## Data Management

Telemetry records are organized using Pandas and stored in
a SQLite database.

The database contains two primary tables:

### Telemetry

Stores:
- Run identifier
- Simulation time
- Timestamp
- Sensor measurements

### Events

Stores:
- Run identifier
- Timestamp
- Event description

## Engineering Concepts Demonstrated

- Object-oriented programming
- Software modularity
- Data acquisition and processing
- Statistical analysis
- Database design and SQL queries
- Fault detection
- System monitoring
- Data visualization
- Automated reporting
- Software integration

## Running the Project

The project was developed in Google Colab.

To run it:

1. Open the notebook in Google Colab.
2. Execute the notebook cells in order.
3. Initialize the simulation environment.
4. Run a virtual test sequence.
5. Explore the monitoring dashboards.
6. Run diagnostics and anomaly detection.
7. Save telemetry to SQLite.
8. Compare test runs.
9. Generate a PDF test report.

## Project Limitations

This is a software-only simulation.

Sensor values, test phases, safety thresholds, and diagnostic
conditions are illustrative. They are not validated engineering
limits and must not be used for real propulsion operations.

The SQLite database is stored in the notebook environment.
Google Colab temporary storage may be cleared when its runtime
is disconnected.

## Future Enhancements

Potential improvements include:

- Real-time MQTT telemetry ingestion
- REST API integration
- Web-based monitoring interface
- Additional sensor simulation models
- More advanced statistical diagnostics
- Automated test-run scheduling
- Cloud database integration
- Unit and integration testing

## Author

Cameron D. Frost

Computer Science

Mississippi State University
"""

filename = "Rocket_Test_Stand_README.md"

with open(filename, "w", encoding="utf-8") as file:
    file.write(project_readme)

print("PROJECT DOCUMENTATION GENERATED")
print("=" * 45)
print("Project: Virtual Rocket Engine Test Stand Simulator")
print("Format: Markdown")
print("File:", filename)
print("Documentation Length:", len(project_readme), "characters")

from google.colab import files
files.download(filename)
```
